# P1 — Load, Clean & Merge

```python
# -- Import Libraries

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# -- 1. LOAD DATA

tickets = pd.read_csv("tickets.csv")
teams = pd.read_csv("teams.csv")

print("Tickets Raw Shape:", tickets.shape)
print("Teams Shape:", teams.shape)

print("\nTicket Data:\n",tickets.head)
print("\nteams Data:\n",teams.head)
```

```python
# -- 2. CHECK DATA TYPES


print("Data Types Before Conversion:\n ",tickets.dtypes)

tickets["ticket_id"] = pd.to_numeric(
    tickets["ticket_id"],
    errors="coerce"
)

tickets["resolution_hours"] = pd.to_numeric(
    tickets["resolution_hours"],
    errors="coerce"
)

tickets["satisfaction"] = pd.to_numeric(
    tickets["satisfaction"],
    errors="coerce"
)


print("\nData Types After Conversion:\n",tickets.dtypes)
```

```python
# -- 3. CHECK DUPLICATES

duplicate_count = tickets.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

print("\nDuplicate Rows:\n",tickets[tickets.duplicated()])
```

```python
# -- 4. REMOVE EXACT DUPLICATE

tickets_clean = tickets.drop_duplicates().copy()

print("Before:", len(tickets))
print("After :", len(tickets_clean))
```

```python
# -- 5. MERGE TICKETS WITH TEAMS

df = tickets_clean.merge(
    teams,
    on="team_id",
    how="left"
)


print("Merged Data:\n",df)

print("\nMerged Shape:", df.shape)
```

```python
# -- 6. DATA VALIDATION / ASSERTIONS

# Data Validation

assert len(df) == 12

assert df["department"].isna().sum() == 0

print("Data validation passed:")
print("Final clean rows:", len(df))
print("Unmatched team IDs:", df["department"].isna().sum())
```

# P2 — Derived Field & Department Analysis 

```python
# -- 7. CREATE BREACH_FLAG

df["breach_flag"] = (
    df["resolution_hours"] > 24
).astype(int)

print("Breach flag created:")
print(df[["ticket_id", "resolution_hours", "breach_flag"]])
```

```python
# -- 8. Department-wise Summary

department_summary = (
    df.groupby("department")
      .agg(
          total_tickets=("ticket_id", "count"),
          breached_tickets=("breach_flag", "sum")
      )
      .reset_index()
)

department_summary["sla_breach_rate"] = (
    department_summary["breached_tickets"]
    / department_summary["total_tickets"]
    * 100
)

department_summary["sla_breach_rate"] = (
    department_summary["sla_breach_rate"].round(2)
)


print("\nDepartment Summary:\n",department_summary)
```

```python
# -- 9. TEAM-WISE BREACH RATE

team_summary = (
    df.groupby("team")
      .agg(
          total_tickets=("ticket_id", "count"),
          breached_tickets=("breach_flag", "sum")
      )
      .reset_index()
)

team_summary["breach_rate"] = (
    team_summary["breached_tickets"]
    / team_summary["total_tickets"]
    * 100
)

team_summary["breach_rate"] = (
    team_summary["breach_rate"].round(2)
)

print("\n Team-wise Summary:\n",team_summary)
```

```python
# -- 10. Highest Breach Rate Team

team_summary = df.groupby('team').agg(
    total_tickets=('ticket_id','count'),
    breached_tickets=('breach_flag','sum')
)

team_summary['breach_rate'] = (
    team_summary['breached_tickets']
    / team_summary['total_tickets']
    * 100
).round(2)

team_summary.sort_values(
    by='breach_rate',
    ascending=False
)
```

# P3 — Chart & Exports

```python
# -- 11. MONTH ORDER

month_order = ["Jan", "Feb", "Mar"]

df["month"] = pd.Categorical(
    df["month"],
    categories=month_order,
    ordered=True
)
```

```python
# -- 12. MONTHLY AVERAGE RESOLUTION HOURS

monthly_avg = (
    df.groupby(
        "month",
        observed=False
    )["resolution_hours"]
    .mean()
    .reset_index()
)

monthly_avg["resolution_hours"] = (
    monthly_avg["resolution_hours"].round(2)
)


print("Monthly Average Resolution Hours:\n",monthly_avg)
```

```python
# -- 13. MONTHLY CHART

plt.figure(figsize=(8, 5))

plt.plot(
    monthly_avg["month"],
    monthly_avg["resolution_hours"],
    marker="o"
)

plt.title(
    "Monthly Average Resolution Hours"
)

plt.xlabel("Month")

plt.ylabel(
    "Average Resolution Hours"
)

plt.tight_layout()

plt.savefig(
    r"C:\Users\Jeel\Desktop\Data_Analysis_SET_B_10861\output\python_chart.png",
    dpi=300
)

plt.show()
```

```python
# -- 14.EXPORTS


df.to_csv(
    r"C:\Users\Jeel\Desktop\Data_Analysis_SET_B_10861\output\clean_data.csv",
    index=False
)

department_summary.to_csv(
    r"C:\Users\Jeel\Desktop\Data_Analysis_SET_B_10861\output\python_summary.csv",
    index=False
)

print("Clean data saved successfully!")
print("Department summary saved successfully!")
```
